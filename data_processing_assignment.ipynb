```python
# Problem 1: Employee Bonus Eligibility

employees = {}

n = int(input("Enter number of employees: "))

for i in range(n):
    name = input("Enter employee name: ")
    score = int(input("Enter performance score: "))
    employees[name] = score

highest_score = max(employees.values())

top_performers = []

for name, score in employees.items():
    if score == highest_score:
        top_performers.append(name)

print("Top Performers Eligible for Bonus:")
print(", ".join(top_performers), "- Score:", highest_score)
```

```python
# Problem 2: Search Keyword Analysis

query = input("Enter search query: ")

query = query.lower()
words = query.split()

word_count = {}

for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print("Keywords searched more than once:")

for word, count in word_count.items():
    if count > 1:
        print(word, ":", count)
```

```python
8# Problem 3: Sensor Data Validation

sensor_readings = []

n = int(input("Enter number of readings: "))

for i in range(n):
    value = int(input(f"Enter reading {i}: "))
    sensor_readings.append(value)

valid_readings = []

for i in range(len(sensor_readings)):
    if sensor_readings[i] % 2 == 0:
        valid_readings.append((i, sensor_readings[i]))

print("Valid Sensor Readings:")
print(valid_readings)
```

```python
#Problem 4: Email Domain Usage

emails = []

n = int(input("Enter number of emails: "))

for i in range(n):
    email = input("Enter email: ")
    emails.append(email)

domain_count = {}

for email in emails:
    domain = email.split("@")[1]

    if domain in domain_count:
        domain_count[domain] += 1
    else:
        domain_count[domain] = 1

total = len(emails)

print("\nDomain Usage Percentage:")

for domain in domain_count:
    percentage = (domain_count[domain] / total) * 100
    print(domain, ":", int(percentage), "%")
```

```python
# Problem 5: Sales Spike Detection (User Input)

sales = []

n = int(input("Enter number of days: "))

for i in range(n):
    value = int(input(f"Enter sales for day {i+1}: "))
    sales.append(value)

average = sum(sales) / len(sales)

print("Average Sales:", average)

for i in range(len(sales)):
    if sales[i] > average * 1.3:
        print("Day", i+1, ":", sales[i])
```

```python
# Problem 6: Duplicate User ID Detection

user_ids = []

n = int(input("Enter number of user IDs: "))

for i in range(n):
    user = input("Enter user ID: ")
    user_ids.append(user)

user_count = {}

for user in user_ids:
    if user in user_count:
        user_count[user] += 1
    else:
        user_count[user] = 1

print("Duplicate User IDs:")

for user in user_count:
    if user_count[user] > 1:
        print(user, "appears", user_count[user], "times")
```
